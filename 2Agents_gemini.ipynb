````python
import os
from autogen import ConversableAgent
from autogen.coding import LocalCommandLineCodeExecutor
# Defining the executor
executor = LocalCommandLineCodeExecutor(
    timeout=10,
    work_dir="coding_workspace",
)

# Agent 1 
code_executor_agent = ConversableAgent(
    name="code_executor_agent",
    llm_config=False,
    code_execution_config={
        "executor": executor,
    },
    human_input_mode="NEVER",
)
````

````python
code_writer_system_message="""
You are a helpful Python developer. 
Write clean, self-contained Python code inside ```python``` blocks to solve the user's task. 
Once you see the code executed successfully and the task is complete, reply with the final answer followed by TERMINATE."""

# Agent 2
code_writer_agent = ConversableAgent(
    "code_writer",
    system_message=code_writer_system_message,
     llm_config={"config_list": [{"model": "gemini-3.5-flash-lite", "api_key":os.getenv("GEMINI_KEY"),"api_type": "google"}]},
    # llm_config={"config_list": [{ "model": "gpt-5-mini", "api_key": os.getenv("OPENAI_API_KEY") }]},
    code_execution_config=False,  # Turn off code execution for this agent.
    max_consecutive_auto_reply=2,
    human_input_mode="NEVER",
)
````

````python
chat_result = code_executor_agent.initiate_chat(
    code_writer_agent, message="write python code alculate the first 15 Fibonacci numbers and save them to a file named 'fibonacci.txt', then print them."
)
````
